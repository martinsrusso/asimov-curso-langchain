```python
"""Notebook module docstring.

This notebook demonstrates creating a `ChatPromptTemplate`, initializing
a chat model, and parsing outputs with `StrOutputParser`.
"""
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnableLambda

from dotenv import load_dotenv
load_dotenv()
```

```python
def deixar_maiusculo(text: str) -> str:
    """Converts the input text to uppercase.

    Args:
        text (str): The input string to be converted.

    Returns:
        str: The uppercase version of the input string.
    """
    return text.upper()
```

```python
prompt = ChatPromptTemplate.from_template("Explique em duas frases sobre o {assunto}")
model = ChatGoogleGenerativeAI(model="gemini-3.6-flash")
parser = StrOutputParser()
passo_extra = RunnableLambda(deixar_maiusculo)
```

```python
pipeline = prompt | model | parser | passo_extra
```

```python
# Invoke pipeline and print the text result (TextAccessor prints as string)
resultado = pipeline.invoke({"assunto": "Inteligência Artificial"})
print(str(resultado))
```
